```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


from statsmodels.tsa.seasonal import STL

from scipy import stats
from scipy import optimize
import seaborn as sns; sns.set() # set default plot styles

import statsmodels.api as sm
# import corner
import time


import statsmodels.api as sm
import statsmodels.formula.api as smf
```

```python
# Load all commune-level dataframes from the parent directory's "data" folder
commune_1_vecs = pd.read_csv("../data/commune_vecs/commune_1_vecs.csv")
commune_2_vecs = pd.read_csv("../data/commune_vecs/commune_2_vecs.csv")
commune_5_vecs = pd.read_csv("../data/commune_vecs/commune_5_vecs.csv")
commune_6_vecs = pd.read_csv("../data/commune_vecs/commune_6_vecs.csv")
commune_7_vecs = pd.read_csv("../data/commune_vecs/commune_7_vecs.csv")
commune_8_vecs = pd.read_csv("../data/commune_vecs/commune_8_vecs.csv")
```

```python


# Load data
df = pd.read_csv("../data/commune_vecs/commune_1_vecs.csv")

# Parse date and extract year/month
df['Date'] = pd.to_datetime(df['YearMonth'], errors='coerce')
df['year'] = df['Date'].dt.year
df['month'] = df['Date'].dt.month

# Drop missing values
df = df.dropna(subset=['mean_d1_ndvi', 'PC1_Score'])

# Initialize summary table
model_results = []
```

```python
# Tranovaho
model_results = []  # Make sure this is initialized before appending

# -----------------------------
# 1. OLS MODEL
# -----------------------------
X = sm.add_constant(df['mean_d1_ndvi'])
y = df['PC1_Score']
ols_model = sm.OLS(y, X).fit()
ols_ci = ols_model.conf_int().loc['mean_d1_ndvi']

print("\n--- OLS Summary: Commune 1 ---")
print(ols_model.summary())

model_results.append({
    'Model Name': 'OLS - Commune 1',
    'NDVI Coeff. Estimate': ols_model.params['mean_d1_ndvi'],
    'Lower Bound': ols_ci[0],
    'Upper Bound': ols_ci[1],
    'P-Value': ols_model.pvalues['mean_d1_ndvi'],
    'R-Squared': ols_model.rsquared,
    'Adjusted R-Squared': ols_model.rsquared_adj,
    'AIC': ols_model.aic,
    'BIC': ols_model.bic
})

# -----------------------------
# 2. LME MODEL (skip appending results)
# -----------------------------
lme_model = smf.mixedlm("PC1_Score ~ mean_d1_ndvi", df, groups=df["month"])
lme_result = lme_model.fit(reml=False, maxiter=2000)

print("\n--- LME Summary: Commune 1 ---")
print(lme_result.summary())

# -----------------------------
# 3. OLS WITH FIXED EFFECTS
# -----------------------------
ols_fe_model = smf.ols("PC1_Score ~ mean_d1_ndvi + C(month)", data=df).fit()
fe_ci = ols_fe_model.conf_int().loc['mean_d1_ndvi']

print("\n--- OLS Summary with Month Fixed Effects ---")
print(ols_fe_model.summary())

model_results.append({
    'Model Name': 'OLS FE - Commune 1',
    'NDVI Coeff. Estimate': ols_fe_model.params['mean_d1_ndvi'],
    'Lower Bound': fe_ci[0],
    'Upper Bound': fe_ci[1],
    'P-Value': ols_fe_model.pvalues['mean_d1_ndvi'],
    'R-Squared': ols_fe_model.rsquared,
    'Adjusted R-Squared': ols_fe_model.rsquared_adj,
    'AIC': ols_fe_model.aic,
    'BIC': ols_fe_model.bic
})

# -----------------------------
# Final Summary Table
# -----------------------------
results_df = pd.DataFrame(model_results)
print("\n--- Summary Table ---")
print(results_df.to_string(index=False))
```
