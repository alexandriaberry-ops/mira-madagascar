```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


from statsmodels.tsa.seasonal import STL

from scipy import stats
from scipy import optimize
import seaborn as sns; sns.set() # set default plot styles

import statsmodels.api as sm
# import corner
import time


import statsmodels.api as sm
import statsmodels.formula.api as smf
```

```python
# Commune-level monthly NDVI and PC1 scores, and the commune lookup table.
# Communes are numbered 1-6 as in the manuscript; data/commune_lookup.csv
# records each commune's name and its original shapefile code.
commune_monthly = pd.read_csv("../../data/commune_monthly.csv")
commune_lookup  = pd.read_csv("../../data/commune_lookup.csv")

commune_name = dict(zip(commune_lookup["commune"], commune_lookup["commune_name"]))
print(commune_lookup)
```

```python


# Commune 2 - Tranovaho
df = commune_monthly[commune_monthly["commune"] == 2].copy()

# Parse date and extract year/month
df['Date'] = pd.to_datetime(df['YearMonth'], errors='coerce')
df['year'] = df['Date'].dt.year
df['month'] = df['Date'].dt.month

# Drop missing values
df = df.dropna(subset=['ndvi', 'PC1_Score'])

# Initialize summary table
model_results = []
```

```python
# Commune 2 - Tranovaho
model_results = []  # Make sure this is initialized before appending

# -----------------------------
# 1. OLS MODEL
# -----------------------------
X = sm.add_constant(df['ndvi'])
y = df['PC1_Score']
ols_model = sm.OLS(y, X).fit()
ols_ci = ols_model.conf_int().loc['ndvi']

print("\n--- OLS Summary: Commune 2 (Tranovaho) ---")
print(ols_model.summary())

model_results.append({
    'Model Name': 'OLS - Commune 2 (Tranovaho)',
    'NDVI Coeff. Estimate': ols_model.params['ndvi'],
    'Lower Bound': ols_ci[0],
    'Upper Bound': ols_ci[1],
    'P-Value': ols_model.pvalues['ndvi'],
    'R-Squared': ols_model.rsquared,
    'Adjusted R-Squared': ols_model.rsquared_adj,
    'AIC': ols_model.aic,
    'BIC': ols_model.bic
})

# -----------------------------
# 2. LME MODEL (skip appending results)
# -----------------------------
lme_model = smf.mixedlm("PC1_Score ~ ndvi", df, groups=df["month"])
lme_result = lme_model.fit(reml=False, maxiter=2000)

print("\n--- LME Summary: Commune 2 (Tranovaho) ---")
print(lme_result.summary())

# -----------------------------
# 3. OLS WITH FIXED EFFECTS
# -----------------------------
ols_fe_model = smf.ols("PC1_Score ~ ndvi + C(month)", data=df).fit()
fe_ci = ols_fe_model.conf_int().loc['ndvi']

print("\n--- OLS Summary with Month Fixed Effects ---")
print(ols_fe_model.summary())

model_results.append({
    'Model Name': 'OLS FE - Commune 2 (Tranovaho)',
    'NDVI Coeff. Estimate': ols_fe_model.params['ndvi'],
    'Lower Bound': fe_ci[0],
    'Upper Bound': fe_ci[1],
    'P-Value': ols_fe_model.pvalues['ndvi'],
    'R-Squared': ols_fe_model.rsquared,
    'Adjusted R-Squared': ols_fe_model.rsquared_adj,
    'AIC': ols_fe_model.aic,
    'BIC': ols_fe_model.bic
})

# -----------------------------
# Final Summary Table
# -----------------------------
results_df = pd.DataFrame(model_results)
print("\n--- Summary Table ---")
print(results_df.to_string(index=False))
```
